The topic I selected for this assessment was the housing market accross the whole island of Ireland. Housing markets also have high-quality data available, making them particularly suitable for this project. I wanted to explore whether there are distinct trends in prices in different areas of Ireland, particulary on either sides of the border in Ireland. There are different policies and governments in each jurisdiction, as well as a different currency. I also wanted to see how different was the rse in house prices compared to inflation in general.

The notebooks folder in this project has the python code used to parse and clean the data, as well as to make exploratory graphs and choropleths. The raw section of the data folder contains the original Republic of Ireland housing data in the PPR_ALL.csv, Northern Ireland house price data in lps-ni-hpi-detailed-q3-2025.xlsx as well as preliminary csvs at various stages of cleaning the data. 

The choropleth.ipynb file is where I made initial choropleth maps for the project. I had to remove the cell outputs for these as the file was too large, but have screenshots of them in the images folder of the e-portfolio.
