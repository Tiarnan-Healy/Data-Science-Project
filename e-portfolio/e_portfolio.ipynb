#### Dashboard Sections

##### Fixed Period Change

This view focuses on fixed-period price changes to provide a clear snapshot of housing market conditions at a specific point in time. By holding the time period constant across all counties, this approach allows spatial comparison without conflating trends from different economic phases.

Using both quarter-on-quarter and year-on-year changes helps distinguish short-term volatility from more persistent trends. The choropleth aslo gives the number of transaction in the period, so for areas with low sales volumes this means we can take into account any local level outliers or high volume transactions like the sale of an apartment block. The option to adjust for inflation ensures that observed growth reflects real price movements rather than general price level changes.

Nominal house prices refer to transaction prices without adjusting for changes in the general price level. Nominal prices indicate how much buyers pay at a given time, they do not account for inflation.

Real house prices adjust nominal prices for inflation, allowing changes in purchasing power to be assessed. Real price analysis is essential for understanding whether housing is becoming genuinely more expensive relative to other goods and incomes, rather than simply reflecting economy-wide price increases.

##### Custom Period Change

Housing market dynamics often do not align neatly with calendar years. The custom period approach allows users to examine price changes between any two quarters, making it possible to analyse economically meaningful periods such as post-crisis recovery phases, pre- and post-Brexit windows, or post-COVID adjustments.

This method avoids arbitrary time boundaries and enables more flexible, hypothesis-driven exploration of housing market behaviour.

##### Inflation vs House Prices 

Comparing house prices directly with inflation addresses whether rising prices represent real asset growth or simply reflect broader inflationary pressures.

Indexing house prices to a common base period allows trends to be compared across jurisdictions and time without being distorted by differences in price levels. Inflation indexing was chosen instead of directly deflating house prices to preserve interpretability and transparency. By setting house prices to an index value of 100 at a chosen base quarter, relative changes over time become immediately comparable across regions and periods. Plotting indexed prices alongside inflation rates makes divergences immediately visible and supports clearer interpretation of real versus nominal growth.

##### County Trends

County-level time series analysis captures local market dynamics that national averages can obscure. By allowing multiple counties to be viewed simultaneously, this approach highlights divergence in price trajectories and recovery patterns across regions. It also enables a user to analyse the data for a localised areas in ways that cannot be done on the larger map interfaces.

Including cumulative transaction volumes alongside prices provides additional context, helping to distinguish between price movements driven by demand pressures and those influenced by supply constraints or reduced market activity.

#### Limitations

While this analysis provides meaningful insight into housing market dynamics, several limitations should be noted:

1. The analysis focuses on prices and transactions and does not explicitly incorporate income or wage data, limiting direct affordability assessment. Getting data that was useful for two jurisdictions proved too difficult.

2. Aggregation at county level may mask intra-county variation, particularly in larger or more diverse regions and in counties with cities. A large number of Republic of Ireland transactions provided just the county and identifying the exact location was not possible.

3. Inflation is measured using CPI, which may not perfectly reflect housing-related cost pressures.

4. Supply-side factors such as new housing completions and planning constraints are not directly modelled.

These limitations suggest directions for future research rather than undermining the validity of the findings.
