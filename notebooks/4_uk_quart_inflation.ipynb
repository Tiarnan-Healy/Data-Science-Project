```python
quarters = {
    "Q1": ["jan", "feb", "mar"],
    "Q2": ["apr", "may", "jun"],
    "Q3": ["jul", "aug", "sep"],
    "Q4": ["oct", "nov", "dec"]
}

rows = []

for _, r in df.iterrows():
    for q, months in quarters.items():
        values = r[months].dropna()
        if len(values) == 0:
            continue

        rows.append({
            "Year": r["Year"],
            "Quarter": int(q[1]),
            "Period": f"{r['Year']} {q}",
            "Inflation_Rate": values.mean()
        })

inflation_q = pd.DataFrame(rows)
```

```python
inflation_q.to_csv("uk_inflation_quarterly.csv", index=False)
```

```python
roi = pd.read_csv("roi_inflation_monthly.csv")
```

```python
roi = roi[["Month", "VALUE"]].copy()
roi.rename(columns={"VALUE": "Inflation_Rate"}, inplace=True)
```

```python
# Extract Year
roi["Year"] = roi["Month"].str.extract(r"(\d{4})").astype(int)

# Extract Month name
roi["Month_Name"] = roi["Month"].str.replace(r"\d{4}\s+", "", regex=True)
```

```python
# convert month names to numbers
month_map = {
    "January": 1, "February": 2, "March": 3,
    "April": 4, "May": 5, "June": 6,
    "July": 7, "August": 8, "September": 9,
    "October": 10, "November": 11, "December": 12
}

roi["Month_Num"] = roi["Month_Name"].map(month_map)
```

```python
roi["Quarter"] = ((roi["Month_Num"] - 1) // 3 + 1).astype(int)
```

```python
roi_q = (
    roi
    .groupby(["Year", "Quarter"])
    .agg(Inflation_Rate=("Inflation_Rate", "mean"))
    .reset_index()
)
```

```python
roi_q["Period"] = (
    roi_q["Year"].astype(str) + " Q" + roi_q["Quarter"].astype(str)
)
```

```python
roi_q = roi_q[["Year", "Quarter", "Period", "Inflation_Rate"]]
roi_q.head()
```

```python
roi_q["Inflation_Rate"] = roi_q["Inflation_Rate"].round(2)
```

```python
roi_q.to_csv("roi_inflation_quarterly.csv", index=False)
```
