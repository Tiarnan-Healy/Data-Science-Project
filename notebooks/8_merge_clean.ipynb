```python
import pandas as pd

all_island = pd.read_csv("../data/quarterly_all_island.csv")
roi_clean = pd.read_csv("../data/roi_quarterly_clean.csv")
```

```python
roi_counties = [
    "Carlow", "Cavan", "Clare", "Cork", "Donegal", "Dublin", "Galway",
    "Kerry", "Kildare", "Kilkenny", "Laois", "Leitrim", "Limerick",
    "Longford", "Louth", "Mayo", "Meath", "Monaghan", "Offaly",
    "Roscommon", "Sligo", "Tipperary", "Waterford",
    "Westmeath", "Wexford", "Wicklow"
]

```

```python
all_island_no_roi = all_island[
    ~all_island["County"].isin(roi_counties)
].copy()
```

```python
quarterly_all_island_clean = pd.concat(
    [all_island_no_roi, roi_clean],
    ignore_index=True
)
```

```python
# Row counts
print("Old all-island:", all_island.shape)
print("New all-island:", quarterly_all_island_clean.shape)

# Check RoI rows
quarterly_all_island_clean[
    quarterly_all_island_clean["County"] == "Dublin"
].head()

```

```python

# Check NI rows still exist
quarterly_all_island_clean[
    quarterly_all_island_clean["County"] == "Belfast"
].head()
```

```python
quarterly_all_island_clean.to_csv(
    "../data/quarterly_all_island.csv",
    index=False
)
```
