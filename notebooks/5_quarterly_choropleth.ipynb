```python
import pandas as pd

quarterly_all_island = pd.read_csv("quarterly_all_island.csv")

print(quarterly_all_island.head())
print(quarterly_all_island.dtypes)
print(quarterly_all_island["County"].nunique())

```

```python
import geopandas as gpd

geo = gpd.read_file("ireland-counties.geojson")

print(geo.head())
print(geo.columns)
```

```python
import geopandas as gpd
import plotly.express as px
fig = px.choropleth(
    quarterly_all_island,
    geojson=geo.__geo_interface__,
    locations="County",
    featureidkey="properties.NAME_1",
    color="avg_price",
    animation_frame="Period",
    color_continuous_scale="Plasma",
    title="Quarterly Average Property Prices (All-Island)",
    hover_name="County",
    hover_data={
        "transaction_count": True,
        "avg_price": ":,.0f",
        "Period": False  # already shown via animation
    },
    labels={
        "avg_price": "Average Price (€)",
        "transaction_count": "Sales"
    }
)

fig.update_geos(fitbounds="locations", visible=False)
fig.update_layout(
    margin=dict(r=0, t=40, l=0, b=0),
    coloraxis_colorbar=dict(title="Average Price (€)")
)

fig.show()

```

```python
fig.write_html("quarterly_all_island_map.html")
```
