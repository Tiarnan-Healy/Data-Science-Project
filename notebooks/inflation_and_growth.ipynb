```python
import pandas as pd

df = pd.read_csv("../data/all_island_quarterly_inflation_clean.csv")

# Sort correctly
df = df.sort_values(["County", "Year", "Quarter"])

# Nominal price growth
df["price_qoq_pct"] = (
    df.groupby("County")["avg_price"]
      .pct_change() * 100
)

df["price_yoy_pct"] = (
    df.groupby("County")["avg_price"]
      .pct_change(4) * 100
)

# Inflation-adjusted growth (real change)
df["real_price_qoq_pct"] = df["price_qoq_pct"] - df["Inflation_Rate"]
df["real_price_yoy_pct"] = df["price_yoy_pct"] - df["Inflation_Rate"]

df.head()
```

```python
df = df.drop(columns=["real_avg_price"], errors="ignore")
```

```python
df.to_csv(
    "../data/all_island_quarterly_inflation_clean.csv",
    index=False
)
```
