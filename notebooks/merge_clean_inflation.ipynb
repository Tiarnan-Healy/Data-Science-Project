```python
import pandas as pd

all_island = pd.read_csv("../data/all_island_quarterly_inflation.csv")
roi_clean = pd.read_csv("../data/roi_quarterly_clean.csv")
```

```python
ni_keywords = [
    "Antrim", "Ards", "Armagh", "Belfast", "Causeway",
    "Derry", "Fermanagh", "Lisburn", "Mid", "Newry"
]

def is_ni(county):
    return any(k.lower() in county.lower() for k in ni_keywords)

all_island["is_ni"] = all_island["County"].apply(is_ni)
```

```python
roi_existing = all_island[~all_island["is_ni"]].copy()
ni_existing  = all_island[all_island["is_ni"]].copy()
```

```python
roi_updated = (
    roi_existing
    .drop(columns=["transaction_count", "avg_price"])
    .merge(
        roi_clean[[
            "County", "Year", "Quarter", "Period",
            "transaction_count", "avg_price"
        ]],
        on=["County", "Year", "Quarter", "Period"],
        how="left"
    )
)
```

```python
all_island_clean = pd.concat(
    [roi_updated, ni_existing.drop(columns=["is_ni"])],
    ignore_index=True
)
```

```python
# No missing prices
assert all_island_clean["avg_price"].notna().all()

# No missing transaction counts
assert all_island_clean["transaction_count"].notna().all()

# Row count unchanged
assert len(all_island_clean) == len(all_island)
```
